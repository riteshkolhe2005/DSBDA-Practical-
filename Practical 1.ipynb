```python
# 1. Import required libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
# 2. Load dataset from open-source CSV file
df = pd.read_csv("StudentPerformance.csv")
```

```python
# 3. Handle missing values

# Check missing values
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Fill missing numerical values with mean
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)
```

```python
# 4. Describe dataset (statistics, variables, dimensions)

# First 5 rows
print("First 5 Rows:")
print(df.head())
```

```python
# Dataset dimensions
print("Shape of Dataset:", df.shape)
```

```python
# Variable names
print("Columns in Dataset:")
print(df.columns)
```

```python
# Data types
print("Data Types:")
print(df.dtypes)
```

```python
# Statistical summary
print("Statistical Summary:")
print(df.describe())
```

```python
# Dataset information
print("Dataset Info:")
print(df.info())
```

```python
# 5. Perform data type conversion
# Example: Convert numerical columns to float if needed

for col in df.select_dtypes(include=['int64']).columns:
    df[col] = df[col].astype(float)
```

```python
print("Updated Data Types After Conversion:")
print(df.dtypes)
```

```python
print("Dataset After Data Type Conversion:")
print(df.head())
```

```python
# 6. Convert categorical variables to numerical

label_encoder = LabelEncoder()

for col in df.select_dtypes(include=['object']).columns:
    df[col] = label_encoder.fit_transform(df[col])
```

```python
# Final processed dataset
print("Processed Dataset:")
print(df.head())
```
