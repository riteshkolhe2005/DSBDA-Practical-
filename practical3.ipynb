```python
import pandas as pd
import numpy as np
```

```python

# 2. Load Iris dataset
df = pd.read_csv("iris.csv")
```

```python
# 3. Display first 5 rows
df.head()
```

```python
# 4. Compute descriptive statistics for numerical columns

# Mean
print("Mean:")
print(df.mean(numeric_only=True))
```

```python
# Median
print("Median:")
print(df.median(numeric_only=True))
```

```python
# Minimum
print("Minimum Values:")
print(df.min(numeric_only=True))
```

```python
# Maximum
print("Maximum Values:")
print(df.max(numeric_only=True))
```

```python
# Variance
print("Variance:")
print(df.var(numeric_only=True))
```

```python
# Standard Deviation
print("Standard Deviation:")
print(df.std(numeric_only=True))
```

```python
# 5. Group data by categorical variable (Species)
grouped_data = df.groupby("Species")
```

```python
# 6. Species-wise statistical comparison
print("Species-wise Mean:")
print(grouped_data.mean(numeric_only=True))
```

```python
print("Species-wise Median:")
print(grouped_data.median(numeric_only=True))
```

```python
print("Species-wise Minimum:")
print(grouped_data.min(numeric_only=True))
```

```python
print("Species-wise Maximum:")
print(grouped_data.max(numeric_only=True))
```

```python
print("Species-wise Variance:")
print(grouped_data.var(numeric_only=True))
```

```python
print("Species-wise Standard Deviation:")
print(grouped_data.std(numeric_only=True))
```

```python
# 7. Full grouped statistical summary
print("Complete Species-wise Statistics:")
print(grouped_data.describe())
```
