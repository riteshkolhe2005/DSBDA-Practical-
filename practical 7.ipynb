```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag
from nltk.tokenize import sent_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
text = "Natural Language Processing is amazing and helps in analyzing text data easily."
```

```python
tokenize_text = sent_tokenize(text)
print(tokenize_text)
```

```python
tokens = word_tokenize(text) #sentence.split(" ")
print("Tokens:", tokens)
```

```python
# POS Tagging
pos = pos_tag(tokens)
print("POS Tags:", pos)
```

```python
# Stopword Removal
filtered = [word for word in tokens if word.lower() not in stopwords.words('english')]
print("After Stopword Removal:", filtered)
```

```python
# Stemming
ps = PorterStemmer()
stemmed = [ps.stem(word) for word in filtered]
print("Stemmed Words:", stemmed)
```

```python
# Lemmatization
lm = WordNetLemmatizer()
lemmatized = [lm.lemmatize(word) for word in filtered]
print("Lemmatized Words:", lemmatized)
```

```python
text = ["Natural Language Processing is amazing and helps in analyzing text data easily.",
        "Machine learning helps in processing and analyzing data."]


vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(text)

words = vectorizer.get_feature_names_out()

tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=words)

print(tfidf_df)
```
