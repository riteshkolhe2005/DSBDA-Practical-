```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from scipy import stats
```

```python
#2. Load academic dataset
df = pd.read_csv("StudentPerformance.csv")
```

```python
#3. Handle missing values and inconsistencies

# Check missing values
print(df.isnull().sum())
```

```python
# Fill missing numerical values with mean
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)
```

```python
# Standardize column names
df.columns = df.columns.str.strip()
```

```python
# 5. Identify outliers using IQR method

# Select numerical columns
numeric_cols = df.select_dtypes(include=[np.number])
```

```python
Q1 = numeric_cols.quantile(0.25)
Q3 = numeric_cols.quantile(0.75)
IQR = Q3 - Q1
```

```python
outliers = ((numeric_cols < (Q1 - 1.5 * IQR)) |
            (numeric_cols > (Q3 + 1.5 * IQR)))
```

```python
print("Outliers using IQR:")
print(outliers.sum())
```

```python
# 6. Identify outliers using Z-score method

z_scores = np.abs(stats.zscore(numeric_cols))
print("Outliers using Z-score:")
print((z_scores > 3).sum(axis=0))
```

```python
# 7. Apply data transformation

# Log transformation
df_log = numeric_cols.apply(lambda x: np.log1p(x))
```

```python
# Normalization using Min-Max Scaling
scaler = MinMaxScaler()
df_normalized = pd.DataFrame(scaler.fit_transform(numeric_cols),
                             columns=numeric_cols.columns)
```

```python
# Standardization using Standard Scaling
standard_scaler = StandardScaler()
df_standardized = pd.DataFrame(standard_scaler.fit_transform(numeric_cols),
                               columns=numeric_cols.columns)
```

```python
# 8. Display transformed data
print("Log Transformed Data:")
print(df_log.head())
print("----------------------------------------------------------------------------")
print("Normalized Data:")
print(df_normalized.head())
print("----------------------------------------------------------------------------")
print("Standardized Data:")
print(df_standardized.head())
```
